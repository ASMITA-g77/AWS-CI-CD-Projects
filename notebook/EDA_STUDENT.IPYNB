```python
numerical_features=[feature for feature in df.columns if df[feature].dtype!='O']
categorical_features=[feature for feature in df.columns if df[feature].dtype=='O']
```

```python
# feature engineering
df['Average']=(df['math_score']+df['reading_score']+df['writing_score'])/3
df['total_score']=df['math_score']+df['reading_score']+df['writing_score']
```

```python
def result(score):
    if score<50:
        return 'Fail'
    else:
        return 'Pass'
```

```python
df['result'] = df['Average'].apply(result)
```

# Multivariate analysis

```python
num_cols = ['math_score','reading_score','writing_score',
            'Average','total_score']

plt.figure(figsize=(8,6))
sns.heatmap(df[num_cols].corr(),
            annot=True,
            cmap='coolwarm')
plt.show()
```

```python
sns.pairplot(df[num_cols])
plt.show()
```

All numerical variables are positively correlated.
As one score increases, the others tend to increase.
Reading and Writing have the strongest relationship.
Students who excel in reading generally excel in writing.
Math is also positively related to Reading and Writing, but the relationship is slightly weaker.
Average and Total Score are perfectly linearly related because Average is directly calculated from Total Score.
No distinct clusters are visible, indicating there are no obvious groups of students based on scores.
A few outliers with very low scores are present but are not numerous enough to significantly affect the overall pattern.


The narrower and more tightly clustered the points are around a straight line, the stronger the linear relationship between the two variables.
